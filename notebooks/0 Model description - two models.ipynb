## Seepage models

### Background

During inlet closures, the estuary loses water to the ocean by seepage through the sand berm.

We estimate seepage from a basin-scale control-volume balance: 

$$
\frac{dV}{dt}
=
Q_r
+ PA(h)
+ O
+ Q_{gw}
\;-\;
\big(S + E\,A(h) + Q_{ov} \big),
$$

 where $V$ is estuary volume, 
$A(h)$ is the water-surface area at stage $h$, 
$Q_r$ is river inflow, 
$P$ is precipitation rate, 
$E$ is evaporation rate, 
$O$ is wave-driven overwash into the estuary, 
$S$ is seepage out through the berm, $Q_{\mathrm{gw}}$ is net groundwater exchange (positive into the estuary), and  $Q_{\mathrm{ov}}$ is overflow/overtopping out across the berm.

Neglecting groundwater exchange and cross-berm overflow during the analyzed closure days, and treating overwash and evaporation as small at daily timescales, the above equation reduces to:


$$
S \;\approx\; Q_r + P\,A(h) -  \frac{dV}{dt},
$$


We apply Darcy's law to model flow through the berm, 
considering the berm as a porous body with cross-sectional flow area $A_b$ and flow length $L_b$ from estuary to ocean. The seepage flux $S$ [$\mathrm{L}^3\,\mathrm{T}^{-1}$] scales with the head difference $\Delta h = h - h_{ocn}$:

$$ S \;=\;  K\, A_b \,\frac{\Delta h}{L_b},$$

where $K$ [$\mathrm{L}\,\mathrm{T}^{-1}$] is the material hydraulic conductivity. 

Considering that $A_b$ and $L_b$ may vary as a function of  estuary water level, we absorb these geometric dependences into an effective berm conductance $C =  K \cdot A_b/L_b$ [$\mathrm{L}^2\,\mathrm{T}^{-1}$].
Seepage is then modeled as: 

$$
S \;=\; C(\cdot)\,\Delta h,
$$

where ${C}(\cdot)$ is a lumped quantity that may vary with estuary water level $h$ or $\Delta h$.

${C}(\cdot)$  combines material properties and berm geometry. 

${C}(\cdot)$ is modeled with a reference $C_\text{ref}$ [$\mathrm{L}^2\,\mathrm{T}^{-1}$]  times a dimensionless factor $f_C$:

$${C} =  C_{ref} f_C(\cdot)$$

The geometric factor (changing geometry of the berm) is captured by the dimensionless factor $f(\cdot)$.


### Two candidate models

1. **Constant ${C}$ (linear fit):** Linear model (constant conductance)
 
This model assumes the seepage cross-sectional area and path length do not change significantly with water level (alternatively, if flow follows an unchanging set of pipes / preferential flow paths).
 
$$
C(\cdot)=  C_L \quad \Rightarrow \quad S=  C_L \Delta h.
$$

- Advantage: simple, convenient for management/operations.


2.  **Nonlinear (offset–power) model on visitor center water level $h_V$:**

The shifted-power model considers that, for a homogeneous sand/gravel barrier, the flow cross sectional area may increase with rising water level. 

$$
C(\cdot)=  C_V\,\zeta_{V}^{\,b_{V}}
\quad\Rightarrow\quad
S \;=\;  C_V\,\zeta_{V}^{\,b_{V}} \Delta h,
$$

where $f_V = \zeta_{V}^{\,b_{V}}$ parameterizes the dependence of $C$ on estuary water level, $h_V$:

$$
\zeta_{V}\equiv\frac{h_{V}+\delta_{V}}{m_{V}}, \quad
\qquad m_{V}=\mathrm{median}(h_{V}) + \delta_{V}
$$

- Advantage: allows the path length and/or cross-sectional area to vary depending on $h_V$

The vertical offset $\delta_V$ represents the seepage path vertical dimension when estuary water level goes to zero.


The exponent $b_{V}$ and normalization $m_{V}$ keep the $\zeta_V$ term unitless.

The linear model corresponds to a nearly level-independent ratio $A_b/L_b$ (or compensating changes in area and path length), while the nonlinear form allows seepage to strengthen or weaken with water-level rise via geometry (wetted thickness, preferential conduits).

Note that $A_b = 0$ may not coincide with $h_V = 0$, because $h_V$ is referenced to NAVD 88 and not estuary depth ($h_V = 0$ m coincides with mean low tide at this location).

**Note: we expect $b<1$**

As estuary level rises, we expect that the active cross‑section $A_b$ (saturated thickness) increases and path length $L_B$ decreases. Thus, $A_b/L_b$ increasese, so the effective conductance increases with water level. 


### Fitting and evaluation
- **Pooled vs. per-event.** We estimate parameters either (i) **pooled** across all closures or (ii) **per-event**, then summarize.  
- **Information criteria.** AIC/BIC for pooled fits with $k=1$ (linear) and $k=3$ ($\delta$, $b$, $C$); for per-event fits, we sum event AIC/BIC (equivalently, sum $-2\log L$ plus penalties).  
- **Goodness of fit.** Report centered $R^2$ and RMSE; provide unweighted and duration-weighted summaries.

*Notes for readers:* The vertical offset $\delta$ is **model-specific** to the $h_{\text{visitor}}$–power model. The median normalization $m_V$ keeps $\zeta_V$ unitless so that $C$ retains interpretable units while the powered term captures geometry-through-level effects.
