## Seepage models

### Background

During inlet closures, the estuary loses water through the wave-built sand berm. 

We use Darcy's law to model flow through the berm, 
considering the berm as a porous body with cross-sectional flow area $A_b$ and flow length $W_b$ between estuary and ocean.

the seepage flux $S$ [L$^3$ T$^{-1}$] scales with the head difference $\Delta h = h_{\text{est}} - h_{\text{ocn}}$:
$$
S \;=\; \tilde K\,A_b\,\frac{\Delta h}{W_b}\,,$$
where $\tilde K$ [L T$^{-1}$] is the material hydraulic conductivity. Because $A_b$ and $W_b$ can change with water level (e.g., wetted thickness grows while flow path shortens/narrows), we absorb these geometric dependences into an effective berm conductance $\, K\,$ [L T$^{-1}$]. 


We model seepage as:
$$
S \;=\; K(\cdot)\,\Delta h,
$$
where $K(\cdot)$ may vary with estuary water level. 


Sensitivity of $A_b$ and $W_b$ -- and by extension $K(\cdot)$ -- to estuary water level is parameterized with an exponent $b$

We do not expect $A_b = 0$ when $h_e =0$, not least because $h_e$ datum is relative to NAVD 88 and not estuary depth. $h_e = 0$ m coincides to mean low tide at this location.


Seepage is observationally inferred each day from a control-volume balance over the basin:
$$
S \;\approx\; Q_r \;-\; \frac{dV}{dt}\,,
$$
after filtering days with wave overwash and treating evaporation as small relative to $Q_r$ during the analyzed closures (see Methods; schematic in the manuscript). This yields paired $(\Delta h,\,S)$ suitable for regression/likelihood fits. 




## Preliminary notes on seepage modeling

Observation. Seepage appears strong relative to noise in the daily aggregates.

### Two candidate models

#### 1) Linear model (constant conductance)

Assume the hydraulically active flow cross‑section does not change much with water level (i.e., the system follows the “same pipe” each time; path length $W$ across the berm and alongshore length $W$ are roughly constant).
$
S \;=\; K\,\Delta h
$

- Use: simple, transparent, and convenient for management/operations.
- Implication: conductance $K$ is treated as constant over the range of $Delta h$ 

#### 2) Offset–power model  $b \approx 0.2$) for a homogeneous sand/gravel berm

As estuary level rises, the saturated thickness (and thus the active cross‑section) increases, so effective conductance grows with head. With flow length $L$ (and width $W$) not changing much,

$
S \;\approx\; K\,\Delta h\, \Delta h + \text{offset})^{\,b}
\quad\text{with } b \approx 1
$
- Use: captures the physics that the path length and/or hydraulically active area depends on $h_e$ 

### Datum/threshold considerations
- Whether you compare estuary level to mean sea level or mean low water mainly adds a constant shift to $Delta h$; the fitted offset absorbs this.  
- In practice this choice does not change conclusions much over the ranges considered; the “general” offset–power form remains appropriate.



### Models
- **Constant $K$ (linear fit):**
$$
K(\cdot)=K \quad\Rightarrow\quad S=K\,\Delta h.
$$

- **Offset–power on visitor center water level:**
$$
K(\cdot)=K\,z_{V}^{\,b_{V}}
\quad\Rightarrow\quad
S \;=\; K\,\Delta h\,z_{V}^{\,b_{V}}.
$$

where $z_{V}$ parameterizes dependence of $K$ on estuary water level,  $h_e$:

$$
z_{V}\equiv\frac{h_{\text{visitor}}+\text{offset}_{V}}{m_{V}}, \quad
\qquad m_{V}=\mathrm{median}(h_{\text{visitor}})
$$

- **Offset–power on $\Delta h$:**
$$
K(\cdot)=K\,z_{\Delta}^{\,b_{\Delta}}
\quad\Rightarrow\quad
S \;=\; K\,\Delta h\,z_{\Delta}^{\,b_{\Delta}}.
$$

where $z_{e}$ parameterizes dependence of $K$ on the water level difference between estuary and ocean, $\Delta h$:

$$
z_{\Delta}\equiv\frac{\Delta h+\text{offset}_{\Delta}}{m_{\Delta}}, \quad 
m_{\Delta}=\mathrm{median}(\Delta h)
$$


Here the exponents $b_{\Delta}$ and $b_{V}$ are model-specific (estimated separately). The normalizations $m_{\Delta}$ and $m_{V}$ keep the powered terms unitless and stabilize estimation while preserving the physical units of $K$.




### Models compared 
We keep $K$/$\tilde K$ in physical units and make any powered term **dimensionless** with a median scale to stabilize estimation:

- **Linear (constant conductance)**
$$
S \;=\; K\,\Delta h\,.
$$

- **Offset–power on head difference ($\Delta h$)**  
Let $m_{\Delta}=\mathrm{median}(\Delta h)$ and define
$$
z_{\Delta}\;=\;\frac{\Delta h+\text{offset}_{\Delta}}{m_{\Delta}}\,.
$$
Then
$$
S \;=\; K\,\Delta h\,z_{\Delta}^{\,b_{\Delta}}\,,
$$
with the guard $\Delta h+\text{offset}_{\Delta}\ge 0$ to keep real powers.

- **Offset–power on estuary level ($h_{\text{visitor}}$)**  
Let $m_V=\mathrm{median}(h_{\text{visitor}})$ and
$$
z_V\;=\;\frac{h_{\text{visitor}}+\text{offset}_V}{m_V}\,.
$$
Then
$$
S \;=\; K\,\Delta h\,z_V^{\,b_V}\,,
$$
with $h_{\text{visitor}}+\text{offset}_V\ge 0$.

> **Interpretation.** The linear model corresponds to a nearly level-independent ratio $A_b/W_b$ (or compensating changes in area and path length), while the two offset–power forms allow seepage to strengthen or weaken with water-level rise via geometry (wetted thickness, preferential conduits) or datum shifts, consistent with the manuscript’s framing of an integrated conductance $\tilde K$ and near-linear $S$–$\Delta h$ behavior in many events. :contentReference[oaicite:2]{index=2}

### Model-comparison approaches
1. *Pooled fit*: one parameter set fit to all closure events.  
2. *Per-event fits:* parameters estimated separately within each event; results summarized across events.

### Evaluation
- *Information criteria:* *sum* event-level AIC/BIC for per-event fits (equivalently, sum $(-2\log L$) and add penalties).  
- *Goodness of fit:* report R^2 and RMSE; provide unweighted and duration-weighted summaries across events.


### Fitting and evaluation
- **Pooled vs. per-event.** We estimate parameters either (i) **pooled** across all closures or (ii) **per-event**, then summarize.  
- **Information criteria.** AIC/BIC for pooled fits with $k=1$ (linear) and $k=3$ (offset, $b$, $K$); for per-event fits, we sum event AIC/BIC (equivalently, sum $-2\log L$ plus penalties).  
- **Goodness of fit.** Report centered $R^2$ and RMSE; provide unweighted and duration-weighted summaries.

*Notes for readers:* The offsets are **model-specific** (one for the $\Delta h$–power model, one for the $h_{\text{visitor}}$–power model). Median normalizations $m_{\Delta}$ and $m_V$ keep $z_{\Delta}$ and $z_V$ unitless so that $K$ retains interpretable units while the powered terms capture gentle geometry-through-level effects. This connects the statistics directly to Darcy physics and observed event hydrodynamics. :contentReference[oaicite:3]{index=3}
